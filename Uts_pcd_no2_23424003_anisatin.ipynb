```python
# ==========================================
# KONVOLUSI CITRA DENGAN LPF, HPF, dan BPF
# ==========================================

import cv2
import numpy as np
import matplotlib.pyplot as plt

# ==========================================
# 1. Membaca gambar
# ==========================================
img = cv2.imread('/content/images.jpg')   # ganti dengan nama gambar anda
img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

# ==========================================
# 2. Definisi Kernel
# ==========================================

# Low Pass Filter (Blur/Reduksi Noise)
kernel_lpf = (1/9) * np.array([
    [1,1,1],
    [1,1,1],
    [1,1,1]
])

# High Pass Filter (Deteksi Tepi)
kernel_hpf = np.array([
    [-1,-1,-1],
    [-1, 8,-1],
    [-1,-1,-1]
])

# Band Pass Filter / Sharpening
kernel_bpf = np.array([
    [ 0,-1, 0],
    [-1, 5,-1],
    [ 0,-1, 0]
])

# ==========================================
# 3. Proses Konvolusi
# ==========================================

lpf = cv2.filter2D(img_rgb, -1, kernel_lpf)
hpf = cv2.filter2D(img_rgb, -1, kernel_hpf)
bpf = cv2.filter2D(img_rgb, -1, kernel_bpf)

# ==========================================
# 4. Menampilkan Hasil
# ==========================================

plt.figure(figsize=(14,8))

# Original
plt.subplot(2,2,1)
plt.imshow(img_rgb)
plt.title("Original")
plt.axis('off')

# Low Pass Filter
plt.subplot(2,2,2)
plt.imshow(lpf)
plt.title("LPF (Blur / Noise Reduction)")
plt.axis('off')

# High Pass Filter
plt.subplot(2,2,3)
plt.imshow(hpf)
plt.title("HPF (Edge Detection)")
plt.axis('off')

# Band Pass Filter
plt.subplot(2,2,4)
plt.imshow(bpf)
plt.title("BPF (Sharpening)")
plt.axis('off')

plt.tight_layout()
plt.show()
```
